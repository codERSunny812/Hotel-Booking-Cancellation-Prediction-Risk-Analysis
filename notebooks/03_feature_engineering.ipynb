```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("../data/hotel_bookings_cleaned.csv")

print("Shape:", df.shape)
df.head()
```

```python
df["total_guests"] = (
    df["adults"] +
    df["children"] +
    df["babies"]
)

print(df["total_guests"].describe())
```

```python
df[[
    "adults",
    "children",
    "babies",
    "total_guests"
]].head(10)
```

```python
df["zero_guest_booking"] = (
    df["total_guests"] == 0
).astype(int)

print(df["zero_guest_booking"].value_counts())
```

```python
df["total_stay_nights"] = (
    df["stays_in_weekend_nights"] +
    df["stays_in_week_nights"]
)

print(df["total_stay_nights"].describe())
```

```python
df["zero_stay_booking"] = (
    df["total_stay_nights"] == 0
).astype(int)

print(df["zero_stay_booking"].value_counts())
```

```python
df["total_previous_bookings"] = (
    df["previous_cancellations"] +
    df["previous_bookings_not_canceled"]
)

print(df["total_previous_bookings"].describe())
```

```python
df["previous_cancellation_rate"] = np.where(
    df["total_previous_bookings"] > 0,
    df["previous_cancellations"] / df["total_previous_bookings"],
    0
)

print(df["previous_cancellation_rate"].describe())
```

```python
df["has_previous_booking"] = (
    df["total_previous_bookings"] > 0
).astype(int)

print(df["has_previous_booking"].value_counts())
```

```python
df["lead_time_group"] = pd.cut(
    df["lead_time"],
    bins=[-1, 7, 30, 90, 180, 365, np.inf],
    labels=[
        "0-7",
        "8-30",
        "31-90",
        "91-180",
        "181-365",
        "365+"
    ]
)

print(df["lead_time_group"].value_counts().sort_index())
```

```python
df["has_special_request"] = (
    df["total_of_special_requests"] > 0
).astype(int)

print(df["has_special_request"].value_counts())
```

```python
df["has_booking_changes"] = (
    df["booking_changes"] > 0
).astype(int)

print(df["has_booking_changes"].value_counts())
```

```python
df["was_on_waiting_list"] = (
    df["days_in_waiting_list"] > 0
).astype(int)

print(df["was_on_waiting_list"].value_counts())
```

```python
df["requires_parking"] = (
    df["required_car_parking_spaces"] > 0
).astype(int)

print(df["requires_parking"].value_counts())
```

```python
df["has_children_or_babies"] = (
    (df["children"] > 0) |
    (df["babies"] > 0)
).astype(int)

print(df["has_children_or_babies"].value_counts())
```

```python
df["adr_per_guest"] = np.where(
    df["total_guests"] > 0,
    df["adr"] / df["total_guests"],
    0
)

print(df["adr_per_guest"].describe())
```

```python
df["has_weekend_stay"] = (
    df["stays_in_weekend_nights"] > 0
).astype(int)

print(df["has_weekend_stay"].value_counts())
```

```python
month_mapping = {
    "January": 1,
    "February": 2,
    "March": 3,
    "April": 4,
    "May": 5,
    "June": 6,
    "July": 7,
    "August": 8,
    "September": 9,
    "October": 10,
    "November": 11,
    "December": 12
}

df["arrival_month_num"] = df["arrival_date_month"].map(month_mapping)

print(df[[
    "arrival_date_month",
    "arrival_month_num"
]].head())
```

```python
def get_season(month):
    if month in [12, 1, 2]:
        return "Winter"
    elif month in [3, 4, 5]:
        return "Spring"
    elif month in [6, 7, 8]:
        return "Summer"
    else:
        return "Autumn"

df["arrival_season"] = df["arrival_month_num"].apply(get_season)

print(df["arrival_season"].value_counts())
```

```python
new_features = [
    "total_guests",
    "zero_guest_booking",
    "total_stay_nights",
    "zero_stay_booking",
    "total_previous_bookings",
    "previous_cancellation_rate",
    "has_previous_booking",
    "lead_time_group",
    "has_special_request",
    "has_booking_changes",
    "was_on_waiting_list",
    "requires_parking",
    "has_children_or_babies",
    "adr_per_guest",
    "has_weekend_stay",
    "arrival_month_num",
    "arrival_season"
]

print("New features created:")
for feature in new_features:
    print("-", feature)
```

```python
print("Final shape:", df.shape)

print("\nMissing values:")
print(df.isna().sum()[df.isna().sum() > 0])

print("\nData types:")
print(df.dtypes)
```
