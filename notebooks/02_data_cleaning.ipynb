```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("../data/hotel_bookings 2.csv")
```

```python
print("shape of the data base is:",df.shape)
```

```python
df_clean = df.copy()

print("Original shape:", df.shape)
print("Cleaning dataset shape:", df_clean.shape)
```

```python
leakage_cols = [
    "reservation_status",
    "reservation_status_date",
    "assigned_room_type"
]

df_clean = df_clean.drop(columns=leakage_cols)

print("Shape after removing leakage columns:", df_clean.shape)
```

```python
# check the null values 

df_clean.isnull().sum()
```

```python
df_clean["children"] = df_clean["children"].fillna(0)
```

```python
#check for the null values in the childen column
df_clean["children"].isna().sum()
```

```python
df_clean["country"] = df_clean["country"].fillna("Unknown")
```

```python
df_clean["country"].isna().sum()
```

```python
df_clean["agent"] = df_clean["agent"].fillna("Unknown")
df_clean["agent"] = df_clean["agent"].astype(str)
```

```python
df_clean["company_missing"] = df_clean["company"].isna().astype(int)
```

```python
df_clean = df_clean.drop(columns=["company"])
```

```python
#preventing the negative adr 

df_clean["adr"] = df_clean["adr"].clip(lower=0)
```

```python
print("Duplicate rows:", df_clean.duplicated().sum())
```

```python
print("Final shape:", df_clean.shape)

print("\nMissing values:")
print(df_clean.isna().sum()[df_clean.isna().sum() > 0])

print("\nDuplicate rows:", df_clean.duplicated().sum())

print("\nData types:")
print(df_clean.dtypes)
```

```python
df_clean[df_clean["adults"] > 10][
    ["adults", "children", "babies", "hotel", "market_segment", "customer_type", "adr", "is_canceled"]
]
```

```python
print("Adults > 10:", (df_clean["adults"] > 10).sum())
print(df_clean.loc[df_clean["adults"] > 10, "adults"].value_counts().sort_index())
```

```python
zero_guests = df_clean[
    (df_clean["adults"] == 0) &
    (df_clean["children"] == 0) &
    (df_clean["babies"] == 0)
]

print("Zero-guest bookings:", len(zero_guests))
zero_guests.head()
```

```python
df_clean = df_clean[df_clean["adults"] <= 10].copy()

print("Shape after removing suspicious adults records:", df_clean.shape)
```

```python
# Remove suspicious records with adults > 10
df_clean = df_clean[df_clean["adults"] <= 10].copy()

print("Final shape:", df_clean.shape)

print("\nMissing values:")
print(df_clean.isna().sum()[df_clean.isna().sum() > 0])

print("\nDuplicate rows:", df_clean.duplicated().sum())

print("\nAdults > 10:", (df_clean["adults"] > 10).sum())

print("\nZero-guest bookings:")
print(
    (
        (df_clean["adults"] == 0) &
        (df_clean["children"] == 0) &
        (df_clean["babies"] == 0)
    ).sum()
)
```

```python
df_clean.to_csv("../data/hotel_bookings_cleaned.csv", index=False)
```
